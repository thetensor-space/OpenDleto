```python
function custom_matrix(n::Int, positions::Vector{Tuple{Int,Int}}, T::Type=Float64, α::T=0, β::T=0, γ::T=0)
    mat = zeros(T, n, n)
    if length(positions) > 0
        mat[positions[1]...] = α
    end
    if length(positions) > 1
        mat[positions[2]...] = β
    end
    if length(positions) > 2
        mat[positions[3]...] = γ
    end
    # Optionally handle up to 6 positions, but only three values provided
    return mat
end
```
