```julia
function relabel_hyperedges(input_file::String, output_file::String)
    txt = read(input_file, String)
    matches = collect(eachmatch(r"\d+", txt))
    all_ints = [parse(Int, m.match) for m in matches]
    unique_ints = sort(unique(all_ints))
    int_map = Dict(x => i for (i, x) in enumerate(unique_ints))
    relabeled_txt = replace(txt, r"\d+" => s -> string(int_map[parse(Int, s)]))
    write(output_file, relabeled_txt)
end

# Example usage:
# relabel_hyperedges("filtered_hyperedges_permuted_3.txt", "filtered_hyperedges_permuted_3_relabel.txt")
```
