Build a diagonal support tensor, then apply invertible changes of coordinates.

```python
import numpy as np
from opendleto import derivations, stratify
rng = np.random.default_rng(42)
base = np.zeros((3, 3, 3))
base[0,0,0], base[1,1,1], base[2,2,2] = 2., 3., 4.
frames = [rng.normal(size=(3,3)) for _ in range(3)]
T = base.copy()
for axis, frame in enumerate(frames):
    T = np.moveaxis(np.tensordot(frame, T, axes=(1, axis)), 0, axis)
```

```python
result = derivations(T, max_vectors=8, tol=1e-8)
print("Exact:", result.exact, "nontrivial dimension:", result.nontrivial_nullity_estimate)
print("Residuals:", result.residuals)
structure = stratify(T, result)
print("Frame method:", structure.method, "allowed positions:", structure.support_count)
print("Forbidden energy:", structure.forbidden_energy, "relative reconstruction error:", structure.reconstruction_error)
print(structure.allowed_support.astype(int))
```

A generic random tensor of the same shape is a negative control.

```python
generic = derivations(rng.normal(size=(3,3,3)))
print("Generic nontrivial dimension:", generic.nontrivial_nullity_estimate)
```
